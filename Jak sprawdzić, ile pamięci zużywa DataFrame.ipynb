```python
import pandas as pd

df = pd.read_csv("movies.csv")
df.head()
```

```python
df.info(memory_usage="deep")
```

```python
mem = df.memory_usage(deep=True)
print("Zużycie pamięci przez kolumny (posortowane):")
print(mem.sort_values(ascending=False))
```

1. **Które kolumny zużywają najwięcej pamięci?**
   - W zbiorze `movies.csv` najwięcej pamięci zużywa kolumna **`crew`** (ok. 19.3 MB) ze względu na bardzo długie opisy techniczne w formie tekstowej. Następne w kolejności są `overview` (1.9 MB) oraz `production_companies`.

2. **Czy to głównie liczby, czy stringi (object)?**
   - Najwięcej pamięci zajmują **stringi (typ object)**. Kolumny liczbowe (np. `budget`, `id`, `popularity`) zajmują zaledwie po ok. 38 KB każda, co stanowi ułamek miejsca zajmowanego przez kolumny tekstowe.
