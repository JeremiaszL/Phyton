```python
import pandas as pd
import numpy as np

def process_sales_optimized(path, chunk_size=50000):
    processed_chunks = []
    
    for chunk in pd.read_csv(path, chunksize=chunk_size):
        
        condition = (chunk["amount"] > 1000) & (chunk["country"] == "PL")
        chunk["flag"] = np.where(condition, 1, 0).astype('int8') # int8 zajmuje tylko 1 bajt
        

        chunk["country"] = chunk["country"].astype("category")
        chunk["product"] = chunk["product"].astype("category")
        
        chunk["amount"] = pd.to_numeric(chunk["amount"], downcast="integer")
        
        processed_chunks.append(chunk)
    
    full_df = pd.concat(processed_chunks, ignore_index=True)
    
    return full_df
```
