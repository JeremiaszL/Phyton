```python
import pandas as pd

def licz_sume_z_podgladem(path, column_name, chunk_size=1000):
    total_sum = 0
    chunk_index = 0
    
    print(f"Rozpoczynam czytanie pliku: {path}\n")
    
    for chunk in pd.read_csv(path, chunksize=chunk_size):
        chunk_index += 1
        
        current_sum = chunk[column_name].sum()
        total_sum += current_sum
        
        print(f"--- Fragment (chunk) nr {chunk_index} ---")
        print(f"Liczba wierszy w tym fragmencie: {len(chunk)}")
        print(f"Suma częściowa ({column_name}) w tym kawałku: {current_sum:,.2f}")
        print(f"Łączna suma do tej pory: {total_sum:,.2f}\n")
        
    print("=" * 40)
    print(f"PRZETWARZANIE ZAKOŃCZONE.")
    print(f"Ostateczna suma kolumny '{column_name}': {total_sum:,.2f}")
    return total_sum

wynik = licz_sume_z_podgladem("movies.csv", "budget", chunk_size=1000)
```
